```python
global_namespace={
    'x':2,
    'f':lambda x:x**2
}
local_namespace={
    'x':3,
    'f':lambda x:x**3
}
eval('x',global_namespace)
```

```python
# eval(expression[, globals[, locals]])
eval('x',local_namespace)
# 按照签名不是不能只传局部不传全局吗 这没报错啊
# 确实，实际上把local_namespace当成全局传进去了，看的是位置不是名字
```

```python
eval('x',global_namespace,local_namespace)
# 局部优先
```

```python
eval('x',local_namespace,global_namespace)
#global_namespace 生效，说明默认就是后面的是局部，前面的是全局
# 只传一个默认是全局
```

```python
eval('f(x)',global_namespace,local_namespace)
#local_namespace生效 
# 函数也是可以的
```
